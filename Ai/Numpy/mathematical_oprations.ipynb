### Vector Operations ---› Operations will be in between the given number of matrices

```python
arr_1 = np.arange(1, 10).reshape(3,3)
arr_1
```

```python
arr_2 = np.arange(10,19).reshape(3,3)
arr_2
```

```python
arr_1 + arr_2
```

```python
arr_2 - arr_1
```

```python
arr_1 * arr_2
```
