## Numpy Arrays Vs Python Sequences
#### Numpy Arrays have a fixed size at creation , unlike Python Lists which grow dynamically. Chaning the size of an ndim array will create a new array and delete the original.
### The elements in a Numpy Array are required toybe of the same data type and thus will be of the same size in memory.
#### Numpy Arrays facilitate advanced Mathematical and other types of operations on large number of data. Typically, such operations are executed more effeiciently with less code than is possible using python's built in sequences.
